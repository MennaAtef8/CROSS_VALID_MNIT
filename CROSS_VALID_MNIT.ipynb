```python
def create_model() :
  model = models.Sequential([
  layers.Conv2D(filters=32,kernel_size=5,strides=1,padding ='same' ,activation= tf.nn.relu,input_shape=(28,28,1)),
  layers.Conv2D(filters=64,kernel_size=(3,3),strides=1,padding ='same' ,activation= tf.nn.relu),
  layers.MaxPool2D(pool_size=(2,2),strides=2 ,padding='valid'),
  layers.Flatten(),
  layers.Dense(units=128,activation=tf.nn.relu),
  layers.Dense(units=10,activation=tf.nn.softmax),
  ])
  model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])

  return model

```

```python
no_folds =5
kfold = KFold(n_splits=no_folds,shuffle=True ,random_state=42)
test_accuracies =[]
fold_no =1
for train_index,test_index in kfold.split(x):
  print(f'Training for fold {fold_no} ...')
  x_train,x_val = x[train_index],x[test_index]
  y_train,y_val = y[train_index],y[test_index]

  model =create_model()

  model.fit(x_train ,y_train ,epochs=10 ,batch_size =64 ,validation_data=(x_val,y_val))
  test_loss ,test_acc =model.evaluate(x_test,y_test)
  test_accuracies.append(test_acc)
  fold_no +=1
avg_test_accuracy =np.mean(test_accuracies)
print(f'Average test accuracy : {avg_test_accuracy}')



```
